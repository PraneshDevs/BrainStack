Hypotheses in Two-Way ANOVA

1. Main Effect of Mentor (Does Factor A affect the mean?)

H₀: All mentors have the same average rating.
μA = μB = μC

H₁: At least one mentor's average rating is different.
μA ≠ μB (i.e., not all means are equal)


2. Main Effect of Mode (Does Factor B affect the mean?)

H₀: Online and offline modes have the same average rating.
μonline = μoffline

H₁: There is a difference in average rating between online and offline.
μonline ≠ μoffline


3. Interaction Effect (Mentor × Mode)
(Do Factors A and B have an interaction effect on the mean?)

H₀: The mode (online/offline) does not matter based on the mentor.

H₁: The mode effect changes based on the mentor.

A Python bootcamp claims that the average quiz score of students trained using the existing method is 75.

A new training approach is introduced.

To evaluate its effectiveness, a random sample of 50 students trained using the new method is selected.

The sample has:
- Mean score = 78
- Population standard deviation = 10

At the 5% level of significance, test whether students trained using the new method performed better than expected.

Level of Significance (α) = 0.05

Null Hypothesis (H₀):
Average marks ≤ 75

Alternative Hypothesis (H₁):
Average marks > 75

Test the claim using a 5% level of significance.

# Step 1: State the Problem Clearly

- Population Average Marks (μ) = 75
- Population Standard Deviation (σ) = 10

New Python Bootcamp:
- Average Marks (x̄) = 78

Level of Significance (α) = 0.05

Null Hypothesis (H₀):

---

## Step 2: Identify the Type of Test

- One-Sample Z-test
- Right-tailed test (because we are testing whether students performed better)

μ = 75

(New method does not improve performance)

---

## Step 4: Compute the Test Statistic (Z-value)

**Z-test Formula:**

\[
z=\frac{\bar{x}-\mu}{\sigma/\sqrt{n}}

\[
z=\frac{78-75}{10/\sqrt{50}}
\]

\[
z=\frac{3}{10/7.071}
\]

\[
z=\frac{3}{1.414}

\[

## Practical Problem :

A company claims that the average time taken to resolve customer tickets is 30 minutes.
A random sample of 40 tickets shows a mean resolution time of 33 minutes.
At the 5% level of significance, test whether the average resolution time has increased.

### H0 : Current process/statment of No Change - There is no significant difference in the avg resoultion time
### H1 : Statement of change/New process - There is a significant difference in the avg resolution time


```python
from scipy import stats
import numpy as np

observations_40_tickets = np.random.normal (loc=33 , scale=5.0, size=40).astype("int")
t_stats,p_value = stats.ttest_1samp(a = observations_40_tickets, popmean=30 ,alternative ='greater')

if p_value < 0.05:
    print("We Reject H0")
else:
    print("We do not reject H0")
```

### Observation
At 5% level of significance , we can reject the null hypothesis and claim that the avg resoultion time has increased

NOTE : Always p-value by default py.program gives for 2 tailed test
